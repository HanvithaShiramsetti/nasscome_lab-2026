USING *args (VARIABLE NUMBER OF POSITIONAL ARGUMENTS)

```python

def print_names(*names):
    print("Names passed:", names)
    for name in names:
        print("Hello", name)

print_names("Alice", "bob", "Charlie")
```

USING **kwargs (VARIABLE NUMBER OF KEYWORD ARGUMENTS)

```python

def print_info(**info):
    print("User Info:")
    for key, value in info.items():
        print(f"{key} = {value}")

print_info(name="Alice", age=25, city="Delhi")
```

lambda functions,filter and map


```python
from functools import reduce

num=[1,2,3,4,5,6,7,8]
square=list(map(lambda x:x**2,num))
print(square)
even_num=list(filter(lambda x:x%2==0,num))
print(even_num)
red=(reduce(lambda x,y:x+y,num))
print(red)
```
