```python
#cleaning data
data = {
    'Name': ['Alice', None, 'Charlie', 'David'],
    'Age': [25, 30, None, 40],
    'Salary': [50000, 60000, 70000, 80000]
}
d=pd.DataFrame(data)
print(d)
```

```python
#drop rows with missing values
print("\ndrop rows with nan values")
print(d.dropna())
```

```python
#fill missing values with specified values
print("\nfill missing vlues with specific values")
print(d.fillna({'Name':'Unknown','age':d['Age'].mean(),'Salary':1500}))
```

**synthetic data creation**

```python
#using random
import numpy as np
import pandas as pd
np.random.seed(1)
no_rows=12
d_syn=pd.DataFrame({
    'Id':np.arange(1,no_rows+1),
    'marks':np.random.randint(0,100,size=no_rows),
    'Passed':np.random.choice(['Pass','Fail'],size=no_rows)
})
print("synthetic data")
print(d_syn)
```

```python
#filter rows based on conditions
print("\nstudents passed")
print(d_syn[d_syn['Passed']=='Pass'])
print("\nstudents failed")
print(d_syn[d_syn['Passed']=='Fail'])
#mean for filtered data
print("\nmean score of passed students")
print(d_syn[d_syn['Passed']=='Pass']['marks'].mean())
print("\n mean of failed students")
print(d_syn[d_syn['Passed']=='Fail']['marks'].mean())
```
