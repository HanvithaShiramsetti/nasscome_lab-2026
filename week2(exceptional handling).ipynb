TRY EXCEPT BLOCK

```python
try:
    x=int(input("enter a number"))
    print(x)
except ValueError:
    print("invalid input")
```

TRY EXCEPT ELSE FINALLY


```python
try:
  n=int(input("enter a number"))
  r=100/n
except ValueError:
  print("invalid input entered")
except ZeroDivisionError:
  print("division by zero is not posssible")
else:
  print("result",r)
finally:
  print("execution is complete")
```

MULTIPLE EXCEPTIONS


```python
try:
   a=int(input("enter a number"))
   b=10/a
except (ValueError,ZeroDivisionError) as e:
   print("an error occured",e)
```

### Custom Exception for Contact Validation

```python
class ContactValidationError(Exception):
    """Custom exception for contact validation errors."""
    def __init__(self, message="Contact validation failed"):
        self.message = message
        super().__init__(self.message)

def validate_contact(contact):
    if not contact.name:
        raise ContactValidationError("Contact name cannot be empty.")
    if not contact.phone:
        raise ContactValidationError("Contact phone number cannot be empty.")
    if not contact.email:
        raise ContactValidationError("Contact email cannot be empty.")
    print(f"Contact '{contact.name}' is valid.")

# Demonstrate the custom exception
try:
    # This contact is valid
    valid_contact = Contact("Charlie Brown", "555-123-4567", "charlie@example.com")
    validate_contact(valid_contact)

    # This contact will cause an error (empty name)
    invalid_contact_name = Contact("", "555-987-6543", "invalid@example.com")
    validate_contact(invalid_contact_name)

except ContactValidationError as e:
    print(f"Validation Error: {e}")

try:
    # This contact will cause an error (empty email)
    invalid_contact_email = Contact("Diana Prince", "555-111-2222", "")
    validate_contact(invalid_contact_email)

except ContactValidationError as e:
    print(f"Validation Error: {e}")
```
